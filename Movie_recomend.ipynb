```python
# now we are trying to extract the gener name form this 
import ast

def extract_names(text):
    try:
        items = ast.literal_eval(text)
        return [item['name'] for item in items]
    except (ValueError, SyntaxError, TypeError):
        return []

merged_df['genres_list'] = merged_df['genres'].apply(extract_names)

merged_df['keywords_list'] = merged_df['keywords'].apply(extract_names)
```

```python
print(merged_df[
    ['original_title', 'genres_list', 'keywords_list', 'overview']
].head())
```

```python
#compaing the features
merged_df['overview'] = merged_df['overview'].fillna('')

merged_df['combined_features'] = (
    merged_df['overview'] + ' ' +
    merged_df['genres_list'].apply(lambda x: ' '.join(x)) + ' ' +
    merged_df['keywords_list'].apply(lambda x: ' '.join(x))
)

print(merged_df[
    ['original_title', 'combined_features']
].head(3))
```

```python
# here we are appling the counter vector
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(
    max_features=5000,
    stop_words='english'
)

feature_matrix = cv.fit_transform(
    merged_df['combined_features']
)

print("Feature matrix shape:", feature_matrix.shape)
```

```python
# calculating the movie similarity using the cosine similarty
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(feature_matrix)

print("Similarity matrix shape:", similarity_matrix.shape)
```

```python
print("Total movies:", len(merged_df))
print("Total movie features:", feature_matrix.shape[1])
print("Similarity between first two movies:",
      similarity_matrix[0][1])
```
